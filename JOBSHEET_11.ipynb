Praktikum 01 - Sequential Search pada Data Angka (praktikum11_seqsearch.py)

```python
def sequential_search(data, target):
  for i in range(len(data)):
    if data[i] == target:
      return f"Data {target} ditemukan di indeks {i}"
  return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(sequential_search(angka, 30))
print(sequential_search(angka, 100))
```

Praktikum 02 - Binary Search pada Data Angka (praktikum11_binsearch.py)

```python
def binary_search(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return f"Data {target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"Data {target} tidak ditemukan"

angka = [10, 20, 30, 40, 50]
print("Data:", angka)
print(binary_search(angka, 40))
print(binary_search(angka, 100))
```

Praktikum 03 - Sequential Search untuk data Nama (praktikum11_seqsearch_nama.py)

```python
def sequential_search_nama(data, target):
  for i in range(len(data)):
    if data[i].lower() == target.lower():
      return f"nama {target} ditemukan di indeks {i}"
  return f"nama {target} tidak ditemukan"

nama = ["Andi", "Budi", "Cici", "Doni"]
print("Data:", nama)
print(sequential_search_nama(nama, "Cici"))
print(sequential_search_nama(nama, "Eka"))
```

Praktikum 04 - Studi Kasus Nilai Mahasiswa dengan Binary Search (praktikum11_binsearch_nilai.py)

```python
def binary_search_nilai(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return f"Nilai {target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"Nilai {target} tidak ditemukan"

nilai = [60, 70, 75, 80, 85, 90]
print("Data nilai:", nilai)
print(binary_search_nilai(nilai, 85))
print(binary_search_nilai(nilai, 65))
```

Praktikum 05 - perbandingan Waktu Sequential vs Binary Search (praktikum11_waktu.py)

```python
import time

def sequential_search(data, target):
  for i in range(len(data)):
    if data[i] == target:
      return i
  return -1

def binary_search(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return mid
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return -1

data = list(range(1, 100001))
target = 99999

start = time.time()
sequential_search(data, target)
end = time.time()
print("Sequential Search waktu:", end - start, "detik")

start = time.time()
binary_search(data, target)
end = time.time()
print("Binary Search waktu:", end - start, "detik")
```

Praktikum 06 - Kombinasi Sorting dan Searching (praktikum11_sortsearch.py)

```python
import random

def insertion_sort(data):
  for i in range(1, len(data)):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

def binary_search(data, target):
  left, right = 0, len(data)-1
  while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
      return f"{target} ditemukan di indeks {mid}"
    elif data[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return f"{target} tidak ditemukan"

data = [random.randint(1, 100) for _ in range(15)]
print("Data acak sebelum diurutkan:", data)
urut = insertion_sort(data)
print("Data setelah diurutkan:", urut)
target = urut[7]
```

3. Variasi Sequential Search

```python
def sequential_search_count(data, target):
  jumlah_perbandingan = 0
  for i in range(len(data)):
    jumlah_perbandingan += 1
    if data[i] == target:
      return f"Data {target} ditemukan di indeks {i}.\n   Total perbandingan: {jumlah_perbandingan}"
  return f"Data {target} tidak ditemukan.\n   Total perbandingan: {jumlah_perbandingan}"

angka = [10, 20, 30, 40, 50, 60, 70, 80]
print(f"Data: {angka}")
print(f"Panjang Data (N): {len(angka)}")

hasil_awal = sequential_search_count(angka, 10)
print(f"Pencarian 10:\n{hasil_awal}\n")

hasil_tengah = sequential_search_count(angka, 30)
print(f"Pencarian 30:\n{hasil_tengah}\n")

hasil_akhir = sequential_search_count(angka, 80)
print(f"Pencarian 80:\n{hasil_akhir}\n")

hasil_gagal = sequential_search_count(angka, 100)
print(f"Pencarian 100:\n{hasil_gagal}")
```

4. Variasi Binary Search

```python
def binary_search_count(data, target):
  left, right = 0, len(data) - 1
  jumlah_langkah = 0

  print(f"\n--- Pencarian {target} dimulai ---")

  while left <= right:
    jumlah_langkah += 1
    mid = (left + right) // 2
    current_value = data[mid]

    print(f"Langkah ke-{jumlah_langkah}: Cek indeks {mid} (Nilai: {current_value})")

    if current_value == target:
      return (f"Data {target} ditemukan di indeks {mid}.\n"
              f"Total langkah (iterasi) yang dibutuhkan: {jumlah_langkah}")
    elif current_value < target:
      left = mid + 1
    else:
      right = mid - 1
  return (f"Data {target} tidak ditemukan.\n"
          f"Total langkah (iterasi) yang dilakukan: {jumlah_langkah}")

angka_terurut = [10, 20, 30, 40, 50, 60, 70, 80, 90]
N = len(angka_terurut)
print(f"Data: {angka_terurut}")
print(f"Panjang Data (N): {N}")

hasil_ditemukan = binary_search_count(angka_terurut, 70)
print(hasil_ditemukan)

hasil_gagal = binary_search_count(angka_terurut, 100)
print(hasil_gagal)
```

5. Studi Kasus Nama Mahasiswa

```python
def sequential_search_nama(daftar_nama, target_nama):
  target_lower = target_nama.lower()

  print(f"\n--- Memulai Sequential Search untuk '{target_nama}' ---")

  for i in range(len(daftar_nama)):
    nama_saat_ini = daftar_nama[i]
    if nama_saat_ini.lower() == target_lower:
      return (f"Nama '{nama_saat_ini}' ditemukan!\n"
              f"Berada di indeks ke-{i} dalam daftar.")
    else:
      print(f"   Membandingkan dengan {nama_saat_ini}... Tidak Cocok.")
  return f"❌ Nama '{target_nama}' tidak ditemukan dalam daftar mahasiswa."

while True:
    try:
        input_string = input("Masukkan daftar nama mahasiswa (pisahkan dengan koma, contoh: Andi, Budi, Cici): ")
        nama_mahasiswa = [nama.strip() for nama in input_string.split(',') if nama.strip()]
        if not nama_mahasiswa:
            print("Daftar tidak boleh kosong. Silakan masukkan nama.")
            continue
        break
    except Exception as e:
        print(f"Terjadi kesalahan input: {e}")

print(f"\n**Daftar Mahasiswa:** {nama_mahasiswa}")

target = input("Masukkan nama mahasiswa yang ingin Anda cari: ")
hasil_pencarian = sequential_search_nama(nama_mahasiswa, target)
print("  ")
print("HASIL AKHIR PENCARIAN")
print("  ")
print(hasil_pencarian)
```

6. Studi Kasus Barang Toko

```python
def binary_search_harga(data, target_harga):
  low = 0
  high = len(data) - 1
  found_items = []

  print(f"\n-- Memulai Binary Search untuk harga {target_harga:,} --")

  while low <= high:
    mid = (low + high) // 2
    harga_mid = data[mid][1]
    if harga_mid == target_harga:
      i = mid
      while i >= 0 and data[i][1] == target_harga:
        if data[i] not in found_items:
             found_items.append(data[i])
        i -= 1
      j = mid + 1
      while j < len(data) and data[j][1] == target_harga:
        if data[j] not in found_items:
             found_items.append(data[j])
        j += 1
      return found_items
    elif harga_mid < target_harga:
      low = mid + 1
    else:
      high = mid - 1
  return []

daftar_barang_mentah = [
    ("Buku Tulis", 7000),
    ("Pensil Mekanik", 25000),
    ("Penghapus", 5000),
    ("Binder A5", 50000),
    ("Stabilo Warna", 15000),
    ("Notes Kecil", 7000),
    ("Kotak Pensil", 35000)
]

print("Daftar Barang Mentah:")
for nama, harga in daftar_barang_mentah:
    print(f"- {nama}: Rp {harga:,}")

print(" " * 50)
barang_terurut = sorted(daftar_barang_mentah, key=lambda item: item[1])

print("Daftar Barang Terurut Berdasarkan Harga:")
for nama, harga in barang_terurut:
    print(f"- {nama}: Rp {harga:,}")

target_harga_cari = 7000
hasil_pencarian = binary_search_harga(barang_terurut, target_harga_cari)

print(f"HASIL PENCARIAN UNTUK HARGA Rp {target_harga_cari:,}")

if hasil_pencarian:
    print(f"Ditemukan {len(hasil_pencarian)} barang dengan harga Rp {target_harga_cari:,}:")
    for nama, harga in hasil_pencarian:
        print(f"  -> {nama}")
else:
    print(f"Tidak ada barang yang ditemukan dengan harga Rp {target_harga_cari:,}.")

target_harga_cari_gagal = 12000
hasil_pencarian_gagal = binary_search_harga(barang_terurut, target_harga_cari_gagal)

print(f"HASIL PENCARIAN UNTUK HARGA Rp {target_harga_cari_gagal:,}")

if not hasil_pencarian_gagal:
    print(f"Tidak ada barang yang ditemukan dengan harga Rp {target_harga_cari_gagal:,}.")
```

7. Analisis Kompleksitas

```python
import math
import time

def sequential_search_worst_case(N):
  return N

def binary_search_worst_case(N):
  if N <= 0:
    return 0
  return math.ceil(math.log2(N + 1))

ukuran_data = [100, 1000, 10000]

print("--- Analisis Kompleksitas Pencarian (Worst Case) ---")
print("{:<10} | {:<20} | {:<20}".format("Ukuran N", "Sequential Search (N)", "Binary Search (log₂ N)"))
print(" ")

tabel_pengamatan = []
for N in ukuran_data:
  seq_ops = sequential_search_worst_case(N)
  bin_ops = binary_search_worst_case(N)

  tabel_pengamatan.append({
      "N": N,
      "Sequential Ops": seq_ops,
      "Binary Ops": bin_ops
  })

  print("{:<10} | {:<20} | {:<20}".format(N, seq_ops, bin_ops))

print(" ")
```

8. Proyek Mini

```python
def binary_search_nilai(data, target):
  low, high = 0, len(data) - 1
  found_students = []

  while low <= high:
    mid = (low + high) // 2
    nilai_mid = data[mid][1]
    if nilai_mid == target:
      i = mid
      while i >= 0 and data[i][1] == target:
        if data[i] not in found_students:
             found_students.append(data[i])
        i -= 1
      j = mid + 1
      while j < len(data) and data[j][1] == target:
        if data[j] not in found_students:
             found_students.append(data[j])
        j += 1
      return sorted(found_students, key=lambda x: x[1])
    elif nilai_mid < target:
      low = mid + 1
    else:
      high = mid - 1
  return []

daftar_mahasiswa_mentah = [
    ("Andi", 85),
    ("Budi", 70),
    ("Cici", 92),
    ("Doni", 70),
    ("Eka", 85),
    ("Fani", 65),
    ("Gita", 92)
]

print("Data Mahasiswa Awal (Tidak Terurut):")
for nama, nilai in daftar_mahasiswa_mentah:
    print(f"- {nama}: Nilai {nilai}")

mahasiswa_terurut = sorted(daftar_mahasiswa_mentah, key=lambda item: item[1])

print("Data Mahasiswa Terurut (Berdasarkan Nilai):")
for nama, nilai in mahasiswa_terurut:
    print(f"- {nama}: Nilai {nilai}")

try:
    target_nilai = int(input("\nMasukkan nilai mahasiswa yang ingin Anda cari: "))
except ValueError:
    print("Input tidak valid. Harap masukkan angka (nilai).")
    exit()

hasil_pencarian = binary_search_nilai(mahasiswa_terurut, target_nilai)

print(f"HASIL PENCARIAN MAHASISWA DENGAN NILAI {target_nilai}")

if hasil_pencarian:
    print(f"Ditemukan {len(hasil_pencarian)} mahasiswa dengan nilai {target_nilai}:")
    for nama, nilai in hasil_pencarian:
        print(f"  -> {nama}")
else:
    print(f"Tidak ada mahasiswa yang ditemukan dengan nilai {target_nilai}.")
```
