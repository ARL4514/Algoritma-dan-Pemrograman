7. Operasi dalam F-String

```python
a = float(input("Masukkan bilangan pertama (A): "))
b = float(input("Masukkan bilangan kedua (B): "))

print("\n--- HASIL ---")
print(f"Hasil Penjumlahan (A + B): {a + b}")
print(f"Hasil Perkalian (A * B): {a * b}")

if b != 0:
    print(f"Hasil Pembagian (A / B): {a / b:.3f}")
else:
    print("Hasil Pembagian (A / B): Tidak terdefinisi (pembagian dengan nol)")
```

8. Casting Data

```python
angka_string = input("Masukkan sebuah angka: ")

try:
    angka_integer = int(angka_string)
    kuadrat = angka_integer ** 2
    angka_float = float(angka_string)

    print("\n--- HASIL KONVERSI ---")
    print(f"Angka sebagai Integer: {angka_integer}")
    print(f"Hasil Kuadrat: {kuadrat}")
    print(f"Angka sebagai Float (2 desimal): {angka_float:.2f}")

except ValueError:
    print("\n⚠️ ERROR: Input tidak valid. Harap masukkan angka yang benar.")
```

9. Studi Kasus - Analisis Teks

```python
def analisis_teks():
    """Membaca kalimat dan melakukan analisis sederhana."""
    kalimat = input("Masukkan sebuah kalimat: ").strip()
    kata_list = kalimat.lower().split()

    if not kata_list:
        print("\nKalimat kosong. Tidak ada data untuk dianalisis.")
        return
    jumlah_kata = len(kata_list)
    kata_terpanjang = max(kata_list, key=len)
    kata_terurut = sorted(kata_list)

    print("\n--- HASIL ANALISIS ---")
    print(f"Kalimat Asli: '{kalimat}'")
    print("-" * 35)
    print(f"1. Jumlah Kata: {jumlah_kata} kata")
    print(f"2. Kata Terpanjang: '{kata_terpanjang}' (Panjang: {len(kata_terpanjang)})")
    print(f"3. Daftar Kata (Urutan Abjad): {kata_terurut}")

analisis_teks()
```
