Praktikum 05 - Operasi pada List 2D

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7, 8, 9], [10, 11, 12]]
C = []

for i in range(len(A)):
  baris = []
  for j in range(len(A[0])):
    baris.append(A[i][j] + B[i][j])
  C.append(baris)

print(C)
```

Praktikum 06 - studi Kasus: Pengelolaan Nilai Mahasiswa

```python
nilai = [
    [80, 90, 85],
    [70, 75, 80],
    [88, 92, 84]
]

for i, baris in enumerate(nilai, start=1):
  rata = sum(baris) / len(baris)
  print(f"Rata-rata mahasiswa {i}: {rata:.2f}")
```

7. Operasi Matriks 2D

```python
matriks_A = [
    [2, 5],
    [1, 6]
]

matriks_B = [
    [4, 3],
    [7, 0]
]

matriks_C = [
    [0, 0],
    [0, 0]
]

for i in range(len(matriks_A)):
  for j in range(len(matriks_A[0])):
    matriks_C[i][j] = matriks_A[i][j] + matriks_B[i][j]

print("Matriks A:")
for row in matriks_A:
  print(row)

print("\nMatriks B:")
for row in matriks_B:
  print(row)

print("\nHasil Penjumlahan (A + B): ")
for row in matriks_C:
  print(row)
```

8. Rata-rata Nilai Mahasiswa

```python
nilai = [
    [75, 80, 85],
    [60, 70, 65],
    [90, 95, 92]
]

for i, nilai_mahasiswa in enumerate(nilai):
    total_nilai = sum(nilai_mahasiswa)
    jumlah_matkul = len(nilai_mahasiswa)
    rata_rata = total_nilai / jumlah_matkul

    print(f"Mahasiswa {i + 1}:")
    print(f"  Nilai: {nilai_mahasiswa}")
    print(f"  Rata-rata: {rata_rata:.2f}")
```

9. Studi Kasus - Data Toko

```python
stok = [
    ["Buku",      20],
    ["Pulpen",    35],
    ["Penghapus", 15],
]

for item in stok:
    nama_barang = item[0]
    jumlah_stok = item[1]
    print(f"{nama_barang}\t\t{jumlah_stok}")

barang_stok_terbanyak = max(stok, key=lambda item: item[1])

nama_terbanyak = barang_stok_terbanyak[0]
stok_terbanyak = barang_stok_terbanyak[1]

print(f"Barang dengan Stok Terbanyak adalah: {nama_terbanyak}")
print(f"Dengan Jumlah Stok: {stok_terbanyak}")
```
