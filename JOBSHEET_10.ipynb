Praktikum 01 - Bubble Sort

```python
def bubble_sort(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] > data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan:", angka)
print("Data setelah Bubble Sort:", bubble_sort(angka))
```

Praktikum 02 - Selection Sort

```python
def selection_sort(data):
  n = len(data)
  for i in range(n):
    min_index = i
    for j in range(i + 1, n):
      if data[j] < data[min_index]:
        min_index = j
      data[i], data[min_index] = data[min_index], data[i]
  return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

Praktikum 03 - Insertion Sort

```python
def insertion_sort(data):
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

Praktikum 04 - Studi Kasus: Mengurutkan Nilai Mahasiswa

```python
def bubble_sort_nilai(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j][1] < data[j + 1][1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

Praktikum 06 - Sorting Data Barang(Nama & Harga)

```python
def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
        else:
          if data[j][key_index] > data[j + 1][key_index]:
            data[j], data[j + 1] = data[j + 1], data[j]
  return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

3. Eksperimen Bubble Sort

```python
def bubble_sort_descending(data):
  n = len(data)
  for i in range(n - 1):
    swapped = False
    for j in range(n - 1 - i):
      if data[j] < data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
        swapped = True
    if not swapped:
      break

data = [64, 34, 25, 12, 22, 11, 90]
print(f"Data sebelum diurutkan: {data}")
bubble_sort_descending(data)
print(f"Data setelah diurutkan (descending): {data}")
```

4. Eksperimen Selection Sort

```python
def selection_sort_largest_to_end(data):
    n = len(data)
    for i in range(n - 1, 0, -1):
        max_idx = 0
        for j in range(1, i + 1):
            if data[j] > data[max_idx]:
                max_idx = j
        data[i], data[max_idx] = data[max_idx], data[i]

data = [64, 34, 25, 12, 22, 11, 90]
print(f"Data sebelum diurutkan: {data}")
selection_sort_largest_to_end(data)
print(f"Data setelah diurutkan (ascending): {data}")
```

```python
import random
def bubble_sort(arr):
    data = arr.copy()
    n = len(data)
    comparisons = 0

    for i in range(n):
        for j in range(0, n-i-1):
            comparisons += 1
            if data[j] > data[j+1]:
                data[j], data[j+1] = data[j+1], data[j]

    return comparisons

def selection_sort(arr):
    data = arr.copy()
    n = len(data)
    comparisons = 0

    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            comparisons += 1
            if data[j] < data[min_index]:
                min_index = j
        data[i], data[min_index] = data[min_index], data[i]

    return comparisons

def insertion_sort(arr):
    data = arr.copy()
    comparisons = 0

    for i in range(1, len(data)):
        key = data[i]
        j = i - 1

        while j >= 0:
            comparisons += 1
            if data[j] > key:
                data[j+1] = data[j]
                j -= 1
            else:
                break

        data[j+1] = key

    return comparisons

sizes = [10, 50, 100]

for size in sizes:
    data = [random.randint(1, 1000) for _ in range(size)]

    b = bubble_sort(data)
    s = selection_sort(data)
    i = insertion_sort(data)

    print(f"Ukuran Input: {size}")
    print(f"Bubble Sort     : {b} perbandingan")
    print(f"Selection Sort  : {s} perbandingan")
    print(f"Insertion Sort  : {i} perbandingan")
```

```python
def input_barang():
    barang_list = []
    jumlah = int(input("Masukkan jumlah barang: "))

    for i in range(jumlah):
        print(f"\nData barang ke-{i+1}")
        nama = input("Nama barang   : ")
        harga = int(input("Harga barang  : "))
        barang_list.append({"nama": nama, "harga": harga})

    return barang_list

def tampilkan_barang(barang_list):
    print("\n=== Daftar Barang ===")
    print("{:<20} {:>10}".format("Nama Barang", "Harga"))
    print("-" * 30)

    for b in barang_list:
        print("{:<20} {:>10}".format(b["nama"], b["harga"]))

def menu_sorting(barang_list):
    print("""
=== Menu Sorting ===
1. Urutkan berdasarkan Nama (Ascending)
2. Urutkan berdasarkan Nama (Descending)
3. Urutkan berdasarkan Harga (Ascending)
4. Urutkan berdasarkan Harga (Descending)
    """)

    pilihan = input("Pilih menu (1/2/3/4): ")

    if pilihan == "1":
        barang_list.sort(key=lambda x: x["nama"])
    elif pilihan == "2":
        barang_list.sort(key=lambda x: x["nama"], reverse=True)
    elif pilihan == "3":
        barang_list.sort(key=lambda x: x["harga"])
    elif pilihan == "4":
        barang_list.sort(key=lambda x: x["harga"], reverse=True)
    else:
        print("Pilihan tidak valid!")

    return barang_list

barang = input_barang()

print("\nData sebelum diurutkan:")
tampilkan_barang(barang)

barang = menu_sorting(barang)

print("\nData setelah diurutkan:")
tampilkan_barang(barang)
```
