Praktikum 01 - Perulangan For dengan Range

```python
for i in range(1, 5):
  print(i)
```

Praktikum 02 - Perulangan While

```python
n = int(input("Masukkan angka: "))
while n > 0:
  print(n)
  n -= 5
```

Praktikum 03 - Nested Loop

```python
for i in range(1, 5):
  for j in range(i):
    print("*", end="")
  print ()
```

Praktikum 04 - Break and Continue

Break

```python
for i in range(1, 10):
  if i == 5:
    print("Angka 5 ditemukan")
    break
```

Continue

```python
for i in range(1, 5):
  if i % 2 == 1:
    continue
  print (i)
```

Praktikum 05 - For pada Dictionary

```python
data = {"Nama": "Andi", "Umur": 20, "Prodi": "TRK"}
for k, v in data.items():
  print(k, ":", v)
```

Praktikum 06 - For dengan Enumerate

```python
daftar = ("Andi", "Budi", "Cici")

for i, nama in enumerate(daftar, start=1):
  print(i, nama)
```

Praktikum 07 - Studi Kasus: Program kasir Multi-Item

```python
total = 0

while True:
  barang =  input("Masukkan nama barang (ketik 'Selesai' untuk berhenti): ")
  if barang.lower() == "selesai":
    break
  harga = float(input(f"Masukkan harga {barang}: "))
  jumlah = int(input(f"Masukkan jumlah {barang}: "))
  subtotal = harga * jumlah
  total += subtotal

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {total}")
```

10. Studi Kasus - Kasir Diskon

```python
def kasir_sederhana():
    """Program kasir sederhana dengan diskon 10% jika total >= Rp 200.000."""
    total_belanja = 0
    print("=== PROGRAM KASIR SEDERHANA ===")
    print("Ketik 'selesai' untuk hitung total.\n")

    while True:
        # Minta input nama barang (atau perintah 'selesai')
        input_nama = input("Nama barang (atau 'selesai'): ").strip().lower()

        if input_nama == 'selesai':
            break

        try:
            # Minta input harga dan kuantitas
            harga = int(input(f"Harga satuan {input_nama}: Rp "))
            qty = int(input(f"Kuantitas {input_nama}: "))

            # Hitung dan tambahkan ke total
            subtotal = harga * qty
            total_belanja += subtotal
            print(f"  Subtotal: Rp {subtotal:,}\n")

        except ValueError:
            print("  ❌ Input harga atau kuantitas harus angka. Coba lagi.\n")

    # --- Perhitungan Diskon ---
    diskon_persen = 0.10  # 10%
    batas_diskon = 200000
    jumlah_diskon = 0

    if total_belanja >= batas_diskon:
        jumlah_diskon = total_belanja * diskon_persen

    total_bayar = total_belanja - jumlah_diskon

    # --- Tampilkan Struk ---
    print("\n\n============================")
    print("      STRUK BELANJA         ")
    print("============================")
    print(f"Total Harga Awal : Rp {total_belanja:,.0f}".replace(",", "."))

    if jumlah_diskon > 0:
        print(f"Diskon (10%)     : Rp {jumlah_diskon:,.0f}".replace(",", "."))
        print("----------------------------")
        print(f"TOTAL BAYAR      : Rp {total_bayar:,.0f}".replace(",", "."))
    else:
        print("Diskon           : Rp 0")
        print("----------------------------")
        print(f"TOTAL BAYAR      : Rp {total_belanja:,.0f}".replace(",", "."))

    print("============================")
    print("   TERIMA KASIH! 👋")

# Jalankan program
kasir_sederhana()
```
